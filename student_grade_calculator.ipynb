```python
# Student Grade Calculator

# Function to calculate grades
def calculate_grade(marks):
    if marks < 0 or marks > 100:
        return "Invalid Marks"
    elif marks >= 90:
        return "A+"
    elif marks >= 80:
        return "A"
    elif marks >= 70:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 50:
        return "D"
    else:
        return "F"


# Student dataset
students = {
    "Ananya": 95,
    "Rahul": 82,
    "Priya": 76,
    "Arjun": 65,
    "Sneha": 52,
    "Kiran": 35
}

# Display student grade report
print("STUDENT GRADE REPORT")
print("-" * 40)
print(f"{'Student Name':<15}{'Marks':<10}{'Grade'}")
print("-" * 40)

for name, marks in students.items():
    grade = calculate_grade(marks)
    print(f"{name:<15}{marks:<10}{grade}")

print("-" * 40)
```

```python
# Test the grading conditions

test_marks = [100, 90, 85, 70, 60, 50, 0, -5, 105]

print("GRADE CALCULATION TEST")
print("-" * 35)

for marks in test_marks:
    grade = calculate_grade(marks)
    print(f"Marks: {marks:<5} Grade: {grade}")

print("-" * 35)
```
